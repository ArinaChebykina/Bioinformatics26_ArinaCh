# Задание 1 - Множественное выравнивание
В качестве исследуемого белка выбран CBG.

**CBG (corticosteroid-binding globulin)** — это транспортный белок плазмы крови, синтезируемый в печени. Его основная функция — связывать и переносить глюкокортикоиды, прежде всего кортизол, в кровотоке. Большая часть кортизола в организме находится в связанном с CBG состоянии, что регулирует его биодоступность и защищает от быстрого разрушения. При снижении уровня CBG увеличивается доля свободного (активного) кортизола.
